## Exercise 1: KL Divergence Between Two Gaussian Distributions


In questo esercizio calcoliamo la divergenza di Kullback-Leibler tra due distribuzioni gaussiane multivariate.

- Nella **prima parte** ricaviamo la formula analitica partendo dalla definizione del KL.
- Poi la implementiamo in Python usando `numpy`.
- Usiamo un esempio concreto con due gaussiane bidimensionali per testare la funzione.
- Infine confrontiamo il valore chiuso con una **stima Monte Carlo** usando campioni.

Questo esercizio serve a comprendere la distanza tra due distribuzioni, concetto centrale nel machine learning probabilistico.


## Exercise 2: Entropy of Nonsingular Linear Transformations


Qui dimostriamo come l'entropia cambia sotto trasformazioni lineari invertibili.

- Mostriamo che se \( \mathbf{y} = A\mathbf{x} \), allora:
  \[
  H[\mathbf{y}] = H[\mathbf{x}] + \log |\det A|
  \]
- È un risultato classico che mostra come l'entropia cresca o decresca in base alla "compressione" operata da \( A \).

Questo esercizio consolida le basi teoriche dell'entropia differenziale e della sua invarianza (o non invarianza) rispetto a trasformazioni.


## Exercise 3: A good reason to go to university


In questo esercizio costruiamo un modello bayesiano per simulare un torneo di tennis tra 4 persone (A, B, C e noi).

- Ogni giocatore ha una skill \( S_i \sim \mathcal{N}(0,1) \)
- Il risultato di ogni match segue una distribuzione Bernoulli con probabilità data da una funzione logistica delle skill.
- Conosciamo gli esiti di due match:
  - A ha vinto contro B
  - B ha vinto contro C
- Stimiamo la probabilità di battere A con campionamento (30.000 simulazioni) in Pyro.

Questo esercizio combina grafi bayesiani, simulazione e uso di modelli probabilistici in Pyro.


## Exercise 4: Bayesian Inference on Carrier Status for Hemophilia


Svolgiamo un'inferenza bayesiana per determinare se Mary è portatrice di emofilia.

- Sappiamo che ha un fratello malato ⇒ la madre è sicuramente portatrice.
- Mary ha 2 figli maschi non affetti ⇒ osservazione usata per aggiornare la credenza.

Analizziamo:
1. La probabilità che Mary sia portatrice a priori e a posteriori.
2. La probabilità che un terzo figlio sia affetto.
3. L'effetto di una ulteriore osservazione (terzo figlio sano) sull'inferenza.

Questo esercizio mostra il potere dell'inferenza bayesiana su modelli genetici reali.
